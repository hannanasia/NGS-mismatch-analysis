## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """Validating individual sequencing reads against reference sequence.
    Converting lowercase nucleotides to uppercase, and identifying
    errors relating to sequence length and invalid characters."""
    sequence = sequence.upper()   #convert sequence to uppercase
    errors = []    #setting up empty errors list
    if len(sequence) != len(reference):   #checking sequence length matches reference
      errors.append("wrong_length")
    for base in sequence:       #checking for invalid characters
      if base not in "ACGT":
        errors.append("invalid_characters")
        break
    valid = len(errors) == 0    #determine which sequence matches reference list and only includes ATGC
    return {                     #dictionary containing uppercase sequence, validity status of sequence, what error (if present)
        "sequence": sequence,
        "valid": valid,
        "errors": errors
    }
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """Identify nucleotide mismatches between a read and the
    reference sequence."""
    mutations = []
    for position, (read_base, reference_base) in enumerate(zip(sequence, reference), start = 1):
        if read_base != reference_base:
            mutations.append({
                "position": position,
                "reference_base": reference_base,
                "read_base": read_base
          })
    return mutations
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """Analysing all sequencing reads, against a reference.
    Validating each read, and identifying mutations in valid reads.
    Return a new list of dictionarise containing each read's normalised
    sequence, validation status, errors, and mutations."""
    results = []
    for read in reads:
        validation = validate_read(read["sequence"], reference)
        if validation["valid"]:
            status = "valid"
            mutations = find_mutations(validation["sequence"], reference)
        else:
            status = "invalid"
            mutations = []
        analysed_read = {
            "read_id": read["read_id"],
            "sample": read["sample"],
            "group": read["group"],
            "sequence": validation["sequence"],
            "status": status,
            "errors": validation["errors"],
            "mutations": mutations
        }
        results.append(analysed_read)

    return results
````

````python
def summarise_samples(analysed_reads):
  """Summarise analysed sequencing reads for each sample and group.
  Counts valid and invalid reads, calculates mismatch percentages,
  identifies the most common mutation positions, and counts substitutions.
  Prints summary and substitution tables for each group."""
  summary = {}      #creating an empty dictionary
  # checking if input list is empty and returning {} if so
  if not analysed_reads:
      return {}
  # looping through each analysed read
  for read in analysed_reads:
        sample = read["sample"]   #assign the sample name to the variable sample
        group = read["group"]   #assign the group (treatment/control) to variable group
        # check if the sample already exists in the summary dictionary
        if sample not in summary:
            # creating a nested dictionary to store results for each sample
            summary[sample] = {
                "group": group,
                "valid_reads": 0,
                "invalid_reads": 0,
                "mismatch_reads": 0,
                "mismatch_percentage": None,
                "mismatch_events": 0,
                "substitution_counts": {},
                "position_counts": {},
                "top_positions": []
            }
        # calculating summary stats for only valid reads
        if read["status"] == "valid":
            summary[sample]["valid_reads"] += 1     #add 1 to valid reads count
        else:
            summary[sample]["invalid_reads"] += 1   #add 1 to invalid reads count
        # looping through valid reads containing mismatch mutations
        if read["status"] == "valid" and read["mutations"]:
            summary[sample]["mismatch_reads"] += 1    #identifying each valid read with at least 1 mutation and adding to mismatch reads
            # for each read with at least 1 mutation +1 to mismatch events for each mutation
            for mutation in read["mutations"]:
                summary[sample]["mismatch_events"] += 1
                substitution = mutation["reference_base"] + ">" + mutation["read_base"]     #identify the nucleotide substitution
                if substitution not in summary[sample]["substitution_counts"]:    #check if substitution already in substitution counts
                    summary[sample]["substitution_counts"][substitution] = 0      # add to dictionary if not and give 0 as initial count
                summary[sample]["substitution_counts"][substitution] += 1   # add 1 to substitution counts
                position = mutation["position"]     #identify mutation position
                if position not in summary[sample]["position_counts"]:    #check if position already in position counts
                    summary[sample]["position_counts"][position] = 0      #set initial count as 0
                summary[sample]["position_counts"][position] += 1     #add 1 to position counts
  # calculating mismatch percentage for valid reads
  for sample in summary:
      if summary[sample]["valid_reads"] > 0:   #check that the sample has valid reads
          #calculating % of valid reads containing mismatches
          summary[sample]["mismatch_percentage"] = (summary[sample]["mismatch_reads"] /
                                                    summary[sample]["valid_reads"]) * 100

      else:
          summary[sample]["mismatch_percentage"] = None     #setting % to None if there are no valid read
      # sorting positions by the number of mismatch events
      sorted_positions = sorted(summary[sample]["position_counts"].items(),
                                key=lambda position: (-position[1], position[0]) )
      # selecting the 3 positions with the most mismatch events
      summary[sample]["top_positions"] = [position_count[0] for position_count in sorted_positions[:3]]


# identify different groups in summary dictionary
  groups = []
  for sample in summary:
      group = summary[sample]["group"]
      if group not in groups:
          groups.append(group)

# table of summarised results
  for group in groups:
      print(f"\nGroup: {group}")      #separate table for each group
      # print column headings
      print(f'{"Sample":<15}{"Valid":>10}{"Invalid":>10}{"Mismatch %":>15}{"Events":>10}{"Top positions":>20}')
      print("-" * 80)
      # print sample summary under its respective group
      for sample in summary:
          if summary[sample]["group"] == group:
              data = summary[sample]
              if data["mismatch_percentage"] is None:
                  percentage = "N/A"    #display N/A for % when the sample has no valid reads
              else:
                  percentage = f'{data["mismatch_percentage"]:.2f}%'  #display % to 2dp
              # print sample results in table
              print(f'{sample:<15}'
                    f'{data["valid_reads"]:>10}'
                    f'{data["invalid_reads"]:>10}'
                    f'{percentage:>15}'
                    f'{data["mismatch_events"]:>10}'
                    f'{str(data["top_positions"]):>20}')
      print()

      # table of mutations
      print("Substitution Mutations:")    #print table heading
      # creating empty list of samples belonging to the current group
      group_samples = []
      for sample in summary:
          if summary[sample]["group"] == group:
              group_samples.append(sample)
      #list all substitutions in respective group
      substitutions = []
      for sample in group_samples:
          for substitution in summary[sample]["substitution_counts"]:
              if substitution not in substitutions:    #adding each substitution once
                  substitutions.append(substitution)
      print(f'{"Substitution":<15}', end="")    #column heading
      for sample in group_samples:
          print(f'{sample:>25}', end="")
      print()   #new line
      print("-" * (15 + 25 * len(group_samples)))   #divider
      for substitution in sorted(substitutions):
          print(f'{substitution:<15}', end="")
          for sample in group_samples:
              count = summary[sample]["substitution_counts"].get(substitution, 0)
              if summary[sample]["valid_reads"] > 0:
                  percentage = (count / summary[sample]["valid_reads"]) * 100
                  result = f'{count} ({percentage:.2f}%)'
              else:
                  result = f'{count} (N/A)'
              print(f'{result:>25}', end="")

          print()

      print()

  return summary
````

## Step 5 — Exploratory comparison of the top mismatch positions

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `sample_summary` is a dictionary in the format returned by the `summarise_samples` function
- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

Reuse helper functions from Step 4 where appropriate to reduce code complexity and repetition.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

### 📋 Return format

Return a list with one dictionary per selected position:

```python
[
    {
        "position": 42,
        "control_n": 3,
        "treated_n": 3,
        "control_mean": 0.5,
        "treated_mean": 2.0,
        "difference": 1.5,
        "statistic": ...,
        "p_value": ...
    }
]
```

Means are sample mismatch percentages; `difference` is **treated mean − control mean**, in percentage points. Use `None` for results that cannot be calculated.

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python
def compare_positions(sample_summary, control_group = "control", treated_group = "treatment"):
    """Compare mismatch percentages at the top 3 positions between groups.
    Uses a two-sided Welch's t-test and returns a list of results for each position."""
    combined_positions = {}   #creating dictionary to store mismatch counts at each position
    results = []    #list to store results for each top position
    for sample in sample_summary:   #loop through each sample in summary dict
        group = sample_summary[sample]["group"]
        if group == control_group or group == treated_group:    #looking at eithe control or treatment
            for position in sample_summary[sample]["position_counts"]:    #mismatch positions for this sample
                count = sample_summary[sample]["position_counts"][position]   #number of mismatches at position
                if position not in combined_positions:    #add position if not already
                    combined_positions[position] = 0
                combined_positions[position] += count   #add sample mismatch count to the combined count
    if combined_positions:    #create list of combined counts
        highest_position = max(combined_positions)
        combined_counts = [0] * highest_position
        for position in combined_positions:
            combined_counts[position - 1] = combined_positions[position]
    else:
        combined_counts = []
    ranked = sorted(combined_positions.items(),
                    key=lambda position_count: (-position_count[1], position_count[0]))   #sorting positions by highest mismatch count
    top_positions = [position_count[0] for position_count in ranked [:3]]
    for sample in sample_summary:
        group = sample_summary[sample]["group"]
        if group == control_group or group == treated_group:
            if sample_summary[sample]["valid_reads"] == 0:
                print(f"Excluding {sample}: no valid reads")   #clearly reporting excluded samples + reason
    # calculating mismatch percentages for each top position
    for position in top_positions:
        control_percentages = []
        treated_percentages = []
        for sample in sample_summary:   #looping thorough each biological sample
            group = sample_summary[sample]["group"]
            valid_reads = sample_summary[sample]["valid_reads"]
            if group == control_group or group == treated_group:
                if valid_reads > 0:
                    count = sample_summary[sample]["position_counts"].get(position, 0)    #mismatch count at position, 0 if no mismatches
                    percentage = (count / valid_reads) * 100    # % per 100 reads
                    if group == control_group:    # adding percentage to the correct group
                        control_percentages.append(percentage)
                    else:
                        treated_percentages.append(percentage)
        if control_percentages:   #calculate mean percentage for control samples
            control_mean = sum(control_percentages) / len(control_percentages)
        else:
            control_mean = None
        if treated_percentages:  #calculate mean percentage for treatment samples
            treated_mean = sum(treated_percentages) / len(treated_percentages)
        else:
            treated_mean = None
        if control_mean is not None and treated_mean is not None:     #calculate differente between T and C means
            difference = treated_mean - control_mean
        else:
            difference = None
        statistical_result = calculate_statistics(control_percentages, treated_percentages)
        statistic = statistical_result["statistic"]
        p_value = statistical_result["p_value"]

        results.append({"position": position,
                        "control_n": len(control_percentages),
                        "treated_n": len(treated_percentages),
                        "control_mean": control_mean,
                        "treated_mean": treated_mean,
                        "difference": difference,
                        "statistic": statistic,
                        "p_value": p_value})


        print(f"\nPosition {position}:")

        if difference is not None:
            print(f"Control mean = {control_mean:.2f}% (n = {len(control_percentages)}), "
                  f"Treatment mean = {treated_mean:.2f}% (n = {len(treated_percentages)}), "
                  f"Difference = {difference:.2f} percentage points")
        else:
            print("Group comparison could not be calculated")

        if statistic is None or p_value is None:
            print("Welch's t-test could not be calculated")
        else:
            print(f"Welch's two-sided t-test: t = {statistic:.3f}, p = {p_value:.3g}")

    return results
````
